```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
iter_array, CSC_array, P1_array, P2_array, P3_array, P4_array = [], [], [], [], [], []
for iter in range (0, 7000, 20):
    file = f"./Sim7/Gen_space_{iter}.txt"
    try:
        print ("Getting file: ", file)
        df = pd.read_csv (file, sep = ' ', header=None)
        iter_array.append(iter)
        headers = ["x", "y", "z", "CSC", "P1", "P2", "P3", "P4", "Act","Nec"]
        df.columns = headers
        CSC_array.append(df["CSC"].sum())
        P1_array.append(df["P1"].sum())
        P2_array.append(df["P2"].sum())
        P3_array.append(df["P3"].sum())
        P4_array.append(df["P4"].sum())
    except FileNotFoundError:
        print(f"File {file} not found. Skipping.")

```

```python
plt.plot (iter_array, CSC_array, label="CSC")
plt.grid (True)
plt.xlabel ("Iteration")
plt.ylabel ("Sum of CSC")
plt.title ("CSC vs Iteration")
plt.legend ()
plt.show ()
```

```python
plt.plot (iter_array, P1_array, label="P1")
plt.grid (True)
plt.xlabel ("Iteration")
plt.ylabel ("Sum of P1")
plt.title ("P1 vs Iteration")
plt.legend ()
plt.show () 
```

```python
plt.plot (iter_array, P2_array, label="P2")
plt.grid (True)
plt.xlabel ("Iteration")
plt.ylabel ("Sum of P2")
plt.title ("P2 vs Iteration")
plt.legend ()
plt.show () 
```

```python
plt.plot (iter_array, P3_array, label="P3")
plt.grid (True)
plt.xlabel ("Iteration")
plt.ylabel ("Sum of P3")
plt.title ("P3 vs Iteration")
plt.legend ()
plt.show () 
```

```python
plt.plot (iter_array, P4_array, label="P4")
plt.grid (True)
plt.xlabel ("Iteration")
plt.ylabel ("Sum of P4")
plt.title ("P4 vs Iteration")
plt.legend ()
plt.show () 
```
